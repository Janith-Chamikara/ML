```python
#look for bigrams
b = {}
for w in words:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1::]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
        
```

```python
import torch
```

```python
N = torch.zeros((27,27),dtype=int)
N
```

```python
print(N.device)
```

```python
chars = sorted(list(set(''.join(words))))
chars
   
```

```python
stoi = {s:i+1 for i, s in enumerate(chars)}
stoi['.'] = 0
stoi
```

```python
for w in words:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1::]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1
        
```

```python
N
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.imshow(N, cmap='gray')
```

```python
itos = {i:s for s, i in stoi.items()}
itos
```

```python
plt.figure(figsize=(16,16))
plt.imshow(N, cmap='Blues')

for i in range(27):
    for j in range(27):
        chstr = itos[i] + itos[j]
        plt.text(j, i,chstr, ha="center", va="bottom")
        plt.text(j, i, N[i, j].item(), ha = 'center',va='top')
```

```python
p = N[0].float()
p = N[0] / p.sum()
p
```

```python
g = torch.Generator().manual_seed(10245678)
ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
itos[ix]
```

```python
P = N.float()
P = P / P.sum(1, keepdim=True)
P[0].sum()
```

```python
g = torch.Generator().manual_seed(10245678)


for i in range(2):
    out = []
    ix = 0
    while True:
        p = P[ix]
        #Uniform distribution - untrained
        # p = torch.ones(27) / 27.0
        ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
        out.append(itos[ix])
        if ix == 0:
            break
        
    print(''.join(out))
```
